```python
import pandas as pd

from chromadb import PersistentClient
from pprint import pprint
from pydantic import BaseModel
```

## Load the dataset

```python
class Book(BaseModel):
    title: str
    author: str
    description: str
```

```python
PATH_TO_DATASET = "data/goodreads_dataset.csv"
df = pd.read_csv(PATH_TO_DATASET)
df = df[df["description"].notna()]
```

```python
books = [Book(**record) for record in df.to_dict(orient="records")]
```

```python
pprint(books)
```

```python
max_description_length = max(len(book.description) for book in books)
print(f"Max description length: {max_description_length}")
```

## Create ChromaDB collection

```python
PATH_TO_CHROMA = "data/"
client  = PersistentClient(path=PATH_TO_CHROMA)
```

```python
chroma_collection = client.get_or_create_collection(name="goodreads-experiments")
```

## Load data

```python
ids = [f"{i + 1}" for i in range(len(books))]
metadatas = [{ "title": b.title, "author": b.author} for b in books]
documents = [b.description for b in books]

chroma_collection.add(ids=ids, metadatas=metadatas, documents=documents)
```
